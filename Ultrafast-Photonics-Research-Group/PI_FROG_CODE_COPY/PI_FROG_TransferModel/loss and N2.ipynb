```python
import numpy as np
import matplotlib.pyplot as plt

# Load the saved values
n2_values = np.load("N2_values.npy")
loss_values = np.load("loss_values.npy")

epochs = np.arange(1000) * 5  # 5 epochs per point

print("finaln2 " + str(n2_values[-1]))
print("finalloss " + str(loss_values[-1]))

print(len(n2_values))
print(len(loss_values))

n2_last_1000 = n2_values[-1000:]
loss_last_1000 = loss_values[-1000:]
epochs_last_1000 = epochs[-1000:]


# Plot N2 values
plt.figure(figsize=(8, 5))
plt.plot(epochs, n2_last_1000, label='N2')
plt.xlabel('Epoch')
plt.ylabel('N2 Value')
plt.title('N2 over Epochs')
plt.grid(True)
plt.legend()
plt.tight_layout()
plt.savefig('N2_plot.png')
plt.show()
plt.close()


clipped_loss_values = np.clip(loss_last_1000, None, 1e2)  # cap at 1000

# Plot Loss values
plt.figure(figsize=(8, 5))
plt.plot(epochs, clipped_loss_values, label='Loss', color='orange')
plt.xlabel('Epoch')
plt.ylabel('Loss')
plt.yscale('log')
plt.title('Loss over Epochs')
plt.grid(True)
plt.legend()
plt.tight_layout()
plt.savefig('Loss_plot.png')
plt.show()
plt.close()
```
